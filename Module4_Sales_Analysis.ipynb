# Module 4 – Python Data & APIs
## Sales Data Analysis Project
This notebook demonstrates NumPy, Pandas, data cleaning, visualization, REST API consumption, JSON processing and report generation.

```python
!pip -q install requests pandas numpy matplotlib seaborn
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import requests

pd.set_option('display.max_columns', None)
```

## 1. Load Dataset

```python
df = pd.read_csv('sales_data.csv')
df.head()
```

## 2. Data Understanding and Cleaning

```python
print('Shape:', df.shape)
print('\nMissing values:\n', df.isnull().sum())
print('\nDuplicate rows:', df.duplicated().sum())

df = df.drop_duplicates()
df['Date'] = pd.to_datetime(df['Date'], errors='coerce')
df['Quantity'] = pd.to_numeric(df['Quantity'], errors='coerce')
df['Unit_Price'] = pd.to_numeric(df['Unit_Price'], errors='coerce')
df['Sales'] = pd.to_numeric(df['Sales'], errors='coerce')
df = df.dropna(subset=['Sales'])
df.describe()
```

## 3. NumPy Basics

```python
sales_array = np.array(df['Sales'])
print('Mean:', np.mean(sales_array))
print('Maximum:', np.max(sales_array))
print('Minimum:', np.min(sales_array))
```

## 4. Pandas Analysis

```python
product_sales = df.groupby('Product')['Sales'].sum().sort_values(ascending=False)
region_sales = df.groupby('Region')['Sales'].sum().sort_values(ascending=False)
category_summary = df.groupby('Category').agg(Total_Sales=('Sales','sum'), Total_Quantity=('Quantity','sum'), Average_Sale=('Sales','mean')).round(2)
print('Product Sales:\n', product_sales)
print('\nRegion Sales:\n', region_sales)
print('\nCategory Summary:\n', category_summary)
```

## 5. Visualization

```python
plt.figure(figsize=(9,5))
product_sales.plot(kind='bar')
plt.title('Sales by Product')
plt.xlabel('Product')
plt.ylabel('Sales (₹)')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
monthly = df.set_index('Date').resample('ME')['Sales'].sum()
plt.figure(figsize=(9,5))
plt.plot(monthly.index, monthly.values, marker='o')
plt.title('Monthly Sales Trend')
plt.xlabel('Month')
plt.ylabel('Sales (₹)')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

## 6. REST API and JSON

```python
# Demo public API – no API key required
url = 'https://jsonplaceholder.typicode.com/users'
response = requests.get(url, timeout=10)
response.raise_for_status()
api_data = response.json()
users = pd.DataFrame(api_data)
users[['id','name','email','company']].head()
```

## 7. Generate Report

```python
top_product = product_sales.idxmax()
top_region = region_sales.idxmax()
report = f'''SALES DATA ANALYSIS REPORT\n\nTotal Sales: ₹{df['Sales'].sum():,.2f}\nTotal Quantity: {df['Quantity'].sum():,.0f}\nOrders: {len(df)}\nAverage Sale: ₹{df['Sales'].mean():,.2f}\nTop Product: {top_product}\nTop Region: {top_region}\n'''
print(report)
with open('sales_analysis_report.txt','w',encoding='utf-8') as f:
    f.write(report)
```
